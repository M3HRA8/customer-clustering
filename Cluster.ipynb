Model_KMEANS

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
inertias = []
sillhouets = []
for k in range(2, 8):
  model_k = KMeans(init='k-means++', n_clusters=k, n_init=12, random_state=42)
  model_k.fit(X)
  inertias.append(model_k.inertia_)
  sillhouets.append(silhouette_score(X, model_k.labels_))

  plt.figure(figsize=(7, 5))
  plt.scatter(feature['Annual Income (k$)'], feature['Spending Score (1-100)'], s=feature['Age']*2, c=model_k.labels_)
  plt.title(f'Annual Income vs Spending Score (k={k})')
  plt.xlabel('Annual Income (k$)')
  plt.ylabel('Spending Score (1-100)')

  centers = model_k.cluster_centers_
  plt.scatter(centers[:, 1], centers[:, 2], c='black', s=300, alpha=0.6)
  plt.show()
  print('='*50)
```

```python
plt.plot(range(2, 8), inertias, 'bo-')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
```

```python
plt.plot(range(2, 8), sillhouets, 'bo-')
plt.title('Silhouette Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
```

#Model_DBSCAN

```python
from sklearn.cluster import DBSCAN
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaler = scaler.fit_transform(X)
X_scaler[:5]
```

```python
plt.scatter(X_scaler[:,1],X_scaler[:,2])
plt.title('Annual Income vs Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
```

```python
model_db = DBSCAN(eps=0.5, min_samples=5)
model_db.fit(X_scaler)
```

```python
labels = model_db.labels_
print(np.sum(labels == -1))
print(len(set(labels)) - (1 if -1 in labels else 0))
```

```python
plt.scatter(X_scaler[:,1],X_scaler[:,2], s=feature['Age']*2, c=model_db.labels_)
plt.title('Annual Income vs Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
from sklearn.neighbors import NearestNeighbors
neighbors = NearestNeighbors(n_neighbors=6)
neighbors.fit(X_scaler)

distances, indices = neighbors.kneighbors(X_scaler)

distances = distances[:, 5]
distances = sorted(distances)

plt.figure(figsize=(8, 5))
plt.plot(distances)

plt.xlabel('Points')
plt.ylabel('Distance')
plt.title('K-Distance Graph')
```

```python
eps_values = [0.5, 0.55, 0.57, 0.58, 0.59, 0.6, 0.65, 0.66, 0.67]

for eps in eps_values:
  model_db = DBSCAN(eps=eps, min_samples=8)
  model_db.fit(X_scaler)
  labels = model_db.labels_
  n_clusters_ = len(set(labels)) - (1 if -1 in labels else 0)
  n_noise_ = np.sum(labels == -1)
  print(f'epsilon: {eps}')
  print('Estimated number of clusters: %d' % n_clusters_)
  print('Estimated number of noise points: %d' % n_noise_)



  plt.scatter(X_scaler[:,1], X_scaler[:,2], c=labels)
  plt.title('Annual Income vs Spending Score')
  plt.xlabel('Annual Income (k$)')
  plt.ylabel('Spending Score (1-100)')
  plt.show()

  print('='*50)
```

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics.pairwise import euclidean_distances
```

```python
X_scaler
```

```python
distances = euclidean_distances(X_scaler)
distances[:5]
```

```python
model_AL = AgglomerativeClustering(n_clusters=5, linkage='ward')
model_AL.fit(X_scaler)
```

```python
model_AL.labels_
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
z = linkage(X_scaler, method='ward')
plt.figure(figsize=(50, 20))
dendrogram(z, leaf_rotation=8)
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distance')
plt.title('Dendrogram')
plt.show()
```

```python
linkages = ['ward', 'complete', 'average', 'single']
sillhouets = []

for lin in linkages:
  model_AL = AgglomerativeClustering(n_clusters=5, linkage=lin)
  model_AL.fit(X_scaler)
  labels = model_AL.labels_
  print(f'linkage: {lin}')
  print('Estimated number of clusters: %d' % len(set(labels)))
  sillhouets.append(silhouette_score(X_scaler, labels))
  print(f'sillhouet score: {sillhouets[-1]}')


  z = linkage(X_scaler, method=lin)
  plt.figure(figsize=(50, 20))
  dendrogram(z, leaf_rotation=8)
  plt.axhline(y=2, linestyle='--')
  plt.xlabel('Data Points')
  plt.ylabel('Euclidean Distance')
  plt.title('Dendrogram')
  plt.show()
  print('=' * 50)

```
